# Simulation des données et analyse statistique des étudiants

Ce notebook utilise **NumPy** pour la génération aléatoire et les statistiques, et **pandas** pour présenter et analyser les données. Il génère 10 étudiants et enregistre `results/report.txt`.

```python
# Simulation des données et analyse statistique des étudiants
# Exécuter ce notebook depuis le dossier student_project.

from pathlib import Path
import pandas as pd
from students import generate_students
from analysis import add_ages, analyze_class, format_report

# Générer 10 étudiants synthétiques (seed fixe pour reproduire le même exemple).
students = generate_students(count=10, seed=42)
students = add_ages(students)

# Afficher les données avec pandas.
df = pd.DataFrame(students)
display(df)

# Calculer puis afficher les statistiques de la classe.
stats = analyze_class(students)
stats_df = pd.DataFrame(
    [{"Statistique": name.replace("_", " ").title(), "Valeur": round(value, 2)}
     for name, value in stats.items()]
)
display(stats_df)

# Créer automatiquement le dossier results et enregistrer le rapport.
results_dir = Path.cwd() / "results"
results_dir.mkdir(parents=True, exist_ok=True)
report_path = results_dir / "report.txt"
report_path.write_text(format_report(students, stats), encoding="utf-8")

print(f"Rapport enregistré dans : {report_path.resolve()}")
print("\n" + report_path.read_text(encoding="utf-8"))
```
