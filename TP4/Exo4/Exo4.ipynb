```python
import pandas as pd
```

```python
# Using the existing hospital DataFrame df,Load it and Display their columns
```

```python
df = pd.read_csv('../Exo3/hospital_original.csv')
```

```python
df
```

```python
#  Add columns
```

```python
df["TotalCost"] =  df["DaysAdmitted"] * df["DailyCost"]
def age_group(age):
    if age < 30:
        return "Young"
    elif age <= 50:
        return "Middle"
    else:
        return "Senior"

df["AgeGroup"] = df["Age"].apply(age_group)
```

```python
df
```

```python
# Add a column for a 10% Bonus based on the daily cost
```

```python
df['Bonus'] = df['DailyCost'] * 0.1
df['Bonus'] = df['Bonus'].astype(int)
```

```python
df
```

```python
# Rename the column DailyCost → DailyFee
```

```python
df = df.rename(columns={"DailyCost": "DailyFee"})
```

```python
# Delete the column Bonus
```

```python
df.drop('Bonus',axis=1,inplace=True)
```

```python
df
```

***Data Transformation:***

```python
# Create a column HighCost showing "Yes" if total cost exceeds 60,000, otherwise "No"
```

```python
def HighCost(TotalCost):
    if TotalCost >60000 :
        return 'Yes'
    else :
        return 'False'

df['HighCost'] = df['TotalCost'].apply(HighCost)
```

```python
# Create a column SatisfactionLevel
```

```python
def SatisfactionLevel(Satisfaction):
    if Satisfaction >= 8 :
        return 'Excellent'
    elif Satisfaction >= 5 :
        return 'Average'
    else :
        return 'Poor'
df['SatisfactionLevel'] = df['Satisfaction'].apply(SatisfactionLevel)
```

```python
df
```

***Ranking and Sorting***

```python
# Add a column ranking patients by TotalCost (highest rank = 1)
```

```python
df['Rank'] = df['TotalCost'].rank(ascending = False).astype(int)
```

```python
# Sort the dataset by RiskScore in descending order
```

```python
df.sort_values(by='Rank')
```

```python
# For each city calculate Average total cost
```

```python
df.groupby('City')['TotalCost',].mean()
```

```python
# For each city calculate Average satisfaction
```

```python
df.groupby('City')['Satisfaction'].mean()
```

```python
# For each city calculate Readmission rate (percentage of patients readmitted)
```

```python
df.groupby('City')['Readmitted'].mean()*100
```

***7. Risk Score Calculation:***

```python
 df['RiskScore'] = (df['DaysAdmitted']*0.4) + (df['Age'] * 0.3) + (100 - df['Satisfaction'] * 10) * 0.3
```

```python
df
```

```python
def risk(risk_score):
    if risk_score > 60:
        return 'High'
    elif risk_score >= 40:
        return 'Medium'
    else :
        return 'Low'
```

```python
df['RiskCategory'] = df['RiskScore'].apply(risk)
```

```python
df
```

```python
#  Find the city with the highest average RiskScore
```

```python
df.groupby('City')['RiskScore'].mean().sort_values(ascending=False).head(1)
```

***8. Apply Data Cleaning and Mapping***

```python
#Replace missing or invalid Satisfaction values with the mean of the column
```

```python
df['Satisfaction'] = df['Satisfaction'].fillna(df['Satisfaction'].mean())
```

```python
df
```

```python
# Map city names alger oran annaba
```

```python
df['City'] = df['City'].replace({'Alger':'DZ-16','Oran':'DZ-31','Annaba':'DZ-23'})
```

```python
df
```

```python
# Replace binary values in Readmitted 
```

```python
df['Readmitted'] = df['Readmitted'].replace({0:'No',1:'Yes'})
```

```python
df
```

***10. Pivot Analysis***

```python
# Create a pivot table showing average TotalCost and Satisfaction for each city and department
```

```python
pd.pivot_table(
    df,
    values=['TotalCost', 'Satisfaction'],
    index=['City','Department'],
    aggfunc='mean'
)
```

```python
city_summary =df.groupby('City').agg(
          AvgTotalCost=('TotalCost', 'mean'),
          AvgSatisfaction=('Satisfaction', 'mean')
)
print(city_summary)
```
