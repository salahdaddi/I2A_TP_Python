```python
import pandas as pd
```

```python
# Using the existing hospital DataFrame df,Load it and Display their columns
```

```python
df = pd.read_csv('../Exo3/hospital_original.csv')
```

```python
df
```

```python
#  Add columns
```

```python
df["TotalCost"] =  df["DaysAdmitted"] * df["DailyCost"]
def age_group(age):
    if age < 30:
        return "Young"
    elif age <= 50:
        return "Middle"
    else:
        return "Senior"

df["AgeGroup"] = df["Age"].apply(age_group)
```

```python
df
```

```python
# Add a column for a 10% Bonus based on the daily cost
```

```python
df['Bonus'] = df['DailyCost'] * 0.1
df['Bonus'] = df['Bonus'].astype(int)
```

```python
df
```

```python
# Rename the column DailyCost → DailyFee
```

```python
df = df.rename(columns={"DailyCost": "DailyFee"})
```

```python
# Delete the column Bonus
```

```python
df.drop('Bonus',axis=1,inplace=True)
```

```python
df
```

***Data Transformation:***

```python
# Create a column HighCost showing "Yes" if total cost exceeds 60,000, otherwise "No"
```

```python
def HighCost(TotalCost):
    if TotalCost >60000 :
        return 'Yes'
    else :
        return 'False'

df['HighCost'] = df['TotalCost'].apply(HighCost)
```

```python
# Create a column SatisfactionLevel
```

```python
def SatisfactionLevel(Satisfaction):
    if Satisfaction >= 8 :
        return 'Excellent'
    elif Satisfaction >= 5 :
        return 'Average'
    else :
        return 'Poor'
df['SatisfactionLevel'] = df['Satisfaction'].apply(SatisfactionLevel)
```

```python
df
```

***Ranking and Sorting***

```python
# Add a column ranking patients by TotalCost (highest rank = 1)
```

```python
df['Rank'] = df['TotalCost'].rank(ascending = False).astype(int)
```

```python
# Sort the dataset by RiskScore in descending order
```

```python
df.sort_values(by='Rank')
```
