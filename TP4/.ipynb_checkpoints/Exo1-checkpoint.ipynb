```python
# 5- Add a new row representing the average temperature of each city over the 7 days.
```

```python
average = np.mean(T[:,1:],axis = 0)
print(average)
avr_with_day = np.concatenate([[T.shape[0] +1], np.around(average,1)])
print(avr_with_day)                              
```

```python
T_new = np.vstack([T,avr_with_day])
print(T_new)
```
