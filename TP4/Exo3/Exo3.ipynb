```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Create a DataFrame

```python
df = pd.DataFrame({
    "City": ["Alger", "Annaba", "Oran", "Alger", "Oran", "Annaba", "Alger", "Oran"],
    "Department": ["Cardiology", "Neurology", "Orthopedics", "Cardiology",
                   "Neurology", "Orthopedics", "Cardiology", "Neurology"],
    "Age": [45, 60, 30, 50, 40, 70, 35, 55],
    "DaysAdmitted": [5, 8, 3, 7, 4, 6, 2, 9],
    "DailyCost": [200, 300, 150, 220, 280, 160, 210, 290],
    "Satisfaction": [4.5, 3.8, 4.2, 4.0, 4.1, 3.5, 4.7, 3.9],
    "Readmitted": [0, 1, 0, 0, 1, 1, 0, 1]
})

print(df)
```

```python
# Create a new DataFrame named df_neurology containing only patients admitted to the
# Neurology department. Keep the columns City, Age, DaysAdmitted and DailyCost, and reset
# the index to start from 0. Then set City as the index and sort the rows by DailyCost in descending order.
```

```python
df_neurology = df[df['Department']=='Neurology'].loc[:,['City','Age','DaysAdmitted','DailyCost','Satisfaction','Readmitted']]
df_neurology.index = np.arange(0,len(df_neurology))
df_neurology = df_neurology.sort_values(by='DailyCost', ascending=False)
df_neurology
```

```python
# Count the number of patients in each city using df_neurology. Display the resulting counts and
```

```python
city_counts = df_neurology['City'].value_counts()
city_counts
```

```python
# create a bar chart to visualize the number of patients per city.
```

```python
plt.bar(city_counts.index,city_counts)
```
