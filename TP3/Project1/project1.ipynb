```python
def clean_lines(list_w):
    clean_list = []
    for word in list_w :
        word = word.strip()
        if word != "" :
            clean_list.append(word)
    return clean_list
    
```

```python
my_list = [" salah"," daddi ","hammou  ","  islaam  "," ryad   ","","\n"]
```

```python
my_list
```

```python
clean_lines(my_list)
```

```python
def parse_products(lines):
    products = []

    for line in lines:
        parts = line.strip().split(",")

        # Check that the line has exactly 3 fields
        if len(parts) != 3:
            continue

        
        name = parts[0].strip()
        price = float(parts[1].strip())
        quantity = int(parts[2].strip())
        products.append((name, price, quantity))

    return products
```

```python
lines = ["Laptop,1000,3", "Phone,500,5", "Tablet,700,2"]
parse_products(lines)
```
