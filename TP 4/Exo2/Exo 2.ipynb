```python
import numpy as np
import matplotlib.pyplot as plt
```

***Data Initialization***

```python
# 1- Generate the grades of 100 students × 5 modules using np.random.randint .
# 2- Fix the random seed with np.random.seed(101) to ensure reproducibility.
# 3- Store the grades in a NumPy array called grades_of_students .
```

```python
help(np.random.randint)
```

```python
grades_of_students = np.random.randint(21,size = (100,5))
np.random.seed(101)
```

```python
# 4- Define the list of modules: ['THL', 'DSS', 'DB', 'SE', 'RE'] 
```

```python
modules = ['THL', 'DSS', 'DB', 'SE', 'RE']
```

```python
# 5- print first five student grades
```

```python
grades_of_students[:5,:]
```

***Basic Statistics***

```python
# 1- Compute the mean grade per student.
```

```python
np.mean(grades_of_students,axis = 1)
```

```python
# 2- Compute the mean grade per module.
```

```python
np.mean(grades_of_students,axis = 0)
```

```python
# 3- Compute the median grade per module.
```

```python
Average = np.median(grades_of_students,axis = 0)
Average
```

```python
# 4- Compute the variance and standard deviation for each module
```

```python
np.var(grades_of_students,axis = 0)
```

```python
np.std(grades_of_students,axis = 0)
```

```python
# 5- Find the overall minimum and maximum grade in the dataset
```

```python
max_grades = np.max(grades_of_students)
min_grades = np.min(grades_of_students)
print(f'maximum grade is : {max_grades}')
print(f'minimum grade is : {min_grades}')
```

```python
# 6- Compute the 25th, 50th, and 75th percentiles for the module THL
```

```python
THL = grades_of_students[:,0]
print(f'the 25th percentiles for the module THL is : {np.percentile(THL,25)}')
print(f'the 50th percentiles for the module THL is : {np.percentile(THL,50)}')
print(f'the 75th percentiles for the module THL is : {np.percentile(THL,75)}')
```

***Boolean Mask and Filtering***

```python
#1- Create a boolean mask indicating grades between 10 and 15.
```

```python
mask = (grades_of_students <= 15) & (grades_of_students >= 10)
```

```python
#2- Count how many grades fall within this range.
```

```python
print(f'grades fall within this range is : {np.sum(mask)}')
```

```python
# 3- Extract all grades between 10 and 15 using the mask.
```

```python
grades_of_students[mask]
```

***Visualization***

```python
# 1-  bar plot to compare the average grades per module
```

```python
plt.bar(modules,Average,edgecolor='black',color = 'orange')
plt.xlabel('Modules')
plt.ylabel('Average')
```

```python
# box plot to show the distribution of grades per module.
```

```python
plt.boxplot(grades_of_students,labels = modules)
plt.title('the distribution of grades per module')
plt.xlabel('Modules')
plt.ylabel('grades_of_students')
```

```python
fig, axes = plt.subplots(1, 5, figsize=(15,4))

for i in range(5):
    axes[i].hist(grades_of_students[:, i],bins=range(0, 22),edgecolor='black',color = 'orange' )
    axes[i].set_title(modules[i])
    axes[i].set_xlabel('Grade')
    axes[i].set_ylabel('Frequency')
plt.show()
```
